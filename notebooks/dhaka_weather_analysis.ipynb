# Dhaka Weather Data Analysis

A Pandas, NumPy and Matplotlib walkthrough of a year of daily Dhaka weather data (synthetic practice data).


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
df = pd.read_csv('../data/dhaka_weather_2025.csv', parse_dates=['date'])
df.head()
```

## 1. Monthly weather summary


```python
monthly = df.groupby('month').agg(
    avg_temperature_c=('temperature_c', 'mean'),
    total_rainfall_mm=('rainfall_mm', 'sum'),
    avg_humidity_percentage=('humidity_percentage', 'mean'),
    avg_air_quality_index=('air_quality_index', 'mean')
).round(2)
monthly
```

## 2. Visualization: temperature trend over the year


```python
monthly_avg = df.set_index('date').resample('M')['temperature_c'].mean()

plt.figure(figsize=(11, 5))
plt.plot(df['date'], df['temperature_c'], alpha=0.45, linewidth=0.8, label='Daily temperature')
plt.plot(monthly_avg.index, monthly_avg.values, color='red', linewidth=2, marker='o', markersize=4, label='Monthly average')
plt.title('Dhaka Daily Temperature Trend - 2025')
plt.xlabel('Date')
plt.ylabel('Temperature (C)')
plt.legend()
plt.show()
```
